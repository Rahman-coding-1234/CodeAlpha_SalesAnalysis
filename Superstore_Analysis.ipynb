```python
df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('Category')['Sales'].sum().sort_values(ascending=False).plot(kind='bar', title='Total Sales by Category')
plt.ylabel('Sales')
plt.show()
```

```python
df.groupby('Order_Year')['Sales'].sum().plot(kind='line', marker='o', title='Yearly Sales Trend')
plt.ylabel('Sales')
plt.show()
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], dayfirst=True)
df['Ship_Date'] = pd.to_datetime(df['Ship_Date'], dayfirst=True)
df['Order_Year'] = df['Order_Date'].dt.year
df['Order_Month'] = df['Order_Date'].dt.month
df[['Order_Date', 'Order_Year', 'Order_Month']].head()
```

```python
df.groupby('Order_Year')['Sales'].sum().plot(kind='line', marker='o', title='Yearly Sales Trend')
plt.ylabel('Sales')
plt.show()
```

```python
yearly = df.groupby('Order_Year')['Sales'].sum()
yearly.index = yearly.index.astype(int)
yearly.plot(kind='line', marker='o', title='Yearly Sales Trend')
plt.ylabel('Sales')
plt.xticks(yearly.index)
plt.show()
```
